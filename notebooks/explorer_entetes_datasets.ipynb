# En-têtes des datasets

Ce notebook parcourt automatiquement `data/raw/` et affiche les colonnes de chaque dataset disponible.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data" / "raw").exists() and (PROJECT_ROOT.parent / "data" / "raw").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
RAW_DIR = PROJECT_ROOT / "data" / "raw"
if not RAW_DIR.exists():
    raise FileNotFoundError(f"Dossier introuvable : {RAW_DIR}")
RAW_DIR
```

```python
FILE_OPTIONS = {
    "Sentiment140.csv": {
        "sep": ",", "encoding": "latin-1", "header": None,
        "names": ["polarity", "tweet_id", "tweet_date", "flag", "user_handle", "text"],
    },
    "dataset_publications.csv": {"sep": ";"},
    "dataset_commentaires.csv": {"sep": ";"},
    "dataset_facebook_joint.csv": {"sep": ";"},
}

def read_header(path: Path) -> list[str]:
    options = {"sep": ",", "encoding": "utf-8", "nrows": 0}
    options.update(FILE_OPTIONS.get(path.name, {}))
    return list(pd.read_csv(path, **options).columns)

rows = []
for path in sorted(RAW_DIR.iterdir()):
    if not path.is_file() or path.name == ".gitkeep":
        continue
    try:
        columns = read_header(path)
        rows.append({"dataset": path.name, "nombre_colonnes": len(columns), "entetes": columns, "statut": "OK"})
    except Exception as error:
        rows.append({"dataset": path.name, "nombre_colonnes": 0, "entetes": [], "statut": f"ERREUR: {error}"})
headers = pd.DataFrame(rows)
display(headers[["dataset", "nombre_colonnes", "statut"]])
```

```python
for _, row in headers.iterrows():
    print(f"\n## {row['dataset']} ({row['nombre_colonnes']} colonnes) — {row['statut']}")
    for position, column in enumerate(row["entetes"], start=1):
        print(f"{position}. {column}")
```
